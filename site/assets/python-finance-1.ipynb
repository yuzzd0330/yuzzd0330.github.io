```python
import pandas as pd

Filepath = r'D:\！Y4 Documents\ACC310\CW2\finaldata.xlsx'
df = pd.read_excel(Filepath)
```

```python
df
```

```python
df = df.fillna(0)
```

```python
df.columns.tolist()
```

```python
df = df[['tick', 'year', 'cashflow_operation', 'net_profit', 'totalassets',
         'inventory', 'cash', 'account_receivable', 'account_payable',
         'shortterm_debt', 'within_oneyear_debt', 'COGS']]
```

```python
df.columns.tolist()
```

```python
tick1 = 408
tick2 = 2450
tick3 = 600518
year_begin = 2015
year_end = 2021

df = df[((df.tick == tick1) | (df.tick == tick2) | (df.tick == tick3)) 
        & (df.year >= year_begin) 
        & (df.year <= year_end)]
```

```python
df[['tick', 'year']]
```

```python
df = df.sort_values(by=['tick', 'year'], ascending=[True, True])
```

```python
df[['tick', 'year']]
```

```python
laglist = ['cashflow_operation', 'net_profit', 'totalassets', 'inventory']

for i in laglist:
    df[i + '_lag'] = df.groupby(['tick'])[i].shift(1)
```

```python
df[['tick', 'year', 'cashflow_operation', 'cashflow_operation_lag']]
```

```python
averagelist = ['totalassets', 'inventory']

for i in averagelist:
    df[i + '_average'] = (df[i] + df[i + '_lag']) / 2
```

```python
df[['tick', 'year', 'totalassets', 'totalassets_lag', 'totalassets_average',
    'inventory', 'inventory_lag', 'inventory_average']]
```

```python
fr = {
    'quick_ratio': {
        'n_add': ['cash', 'account_receivable', 'inventory'],
        'n_sub': [],
        'd_add': ['account_payable', 'shortterm_debt', 'within_oneyear_debt'],
        'd_sub': []
    },

    'cfo_growth': {
        'n_add': ['cashflow_operation'],
        'n_sub': ['cashflow_operation_lag'],
        'd_add': ['cashflow_operation_lag'],
        'd_sub': []
    },

    'net_profit_growth': {
        'n_add': ['net_profit'],
        'n_sub': ['net_profit_lag'],
        'd_add': ['net_profit_lag'],
        'd_sub': []
    },

    'ROA': {
        'n_add': ['net_profit'],
        'n_sub': [],
        'd_add': ['totalassets_average'],
        'd_sub': []
    },

    'account_receivable_percentage': {
        'n_add': ['account_receivable'],
        'n_sub': [],
        'd_add': ['totalassets'],
        'd_sub': []
    },

    'inventory_turnover': {
        'n_add': ['COGS'],
        'n_sub': [],
        'd_add': ['inventory_average'],
        'd_sub': []
    }
}
```

```python
for key, values in fr.items():
    for subkey, subvalues in values.items():
        df[key + subkey] = df[subvalues].sum(axis=1, skipna=False)

    df[key] = (df[key + 'n_add'] - df[key + 'n_sub']) / (df[key + 'd_add'] - df[key + 'd_sub'])
```

```python
df[['tick', 'year', 'quick_ratio', 'cfo_growth', 'net_profit_growth',
    'ROA', 'account_receivable_percentage', 'inventory_turnover']]
```

```python
df = df[(df.tick == 600518) & (df.year >= 2016) & (df.year <= 2021)]
```

```python
df[['tick', 'year']]
```

```python
df = df[['tick', 'year', 'quick_ratio', 'cfo_growth', 'net_profit_growth',
         'ROA', 'account_receivable_percentage', 'inventory_turnover']]
```

```python
round_list = ['quick_ratio', 'cfo_growth', 'net_profit_growth',
              'ROA', 'account_receivable_percentage', 'inventory_turnover']

df[round_list] = df[round_list].round(3)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from matplotlib.ticker import PercentFormatter
```

```python
plt.figure(figsize=(8, 6))

plt.plot(df['year'], df['ROA'], marker='*', color='b')
plt.xticks(df['year'])

plt.title('ROA')
plt.xlabel('Year')
plt.ylabel('ROA')

ax = plt.gca()
ax.yaxis.set_major_formatter(PercentFormatter(1))

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.plot(df['year'], df['cfo_growth'], marker='*', color='b', label='CFO Growth')
plt.plot(df['year'], df['net_profit_growth'], marker='.', color='g', label='Net Profit Growth')
plt.xticks(df['year'])

ax = plt.gca ()
ax.yaxis.set_major_formatter(PercentFormatter(1)) 

plt.title('Growth')
plt.xlabel('Year')
plt.ylabel('Growth Rate')
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 6))

line1, = plt.plot(df['year'], df['quick_ratio'], marker='*', color='b', label='Quick Ratio')
plt.xticks(df['year'])
plt.title('Financial Ratios')
plt.xlabel('Year')
plt.ylabel('Quick Ratio')

ax2 = plt.twinx()
line2, = ax2.plot(df['year'], df['account_receivable_percentage'], marker='.', color='g', label='Account Receivable Percentage')
ax2.set_ylabel('Account Receivable Percentage')

ax2.set_ylim(0.05, 0.09)
ax2.yaxis.set_major_formatter(PercentFormatter(xmax=1)) 

lines = [line1, line2]
labels = [l.get_label() for l in lines]
plt.legend(lines, labels, loc='upper right')

plt.show()
```
